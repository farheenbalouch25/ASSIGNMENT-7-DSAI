Make a class called Employee. The init() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

```python
class Employee:
    def __init__(self, name, number, address, salary, post, joining_year, education, experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.joining_year = joining_year
        self.education = education
        self.experience = experience 
    def describe_employee(self):
        print("\nEmployee Details")
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.joining_year)
        print("Education:", self.education)
        print("Experience:", self.experience)
    def update_year_of_experience(self,years):
        self.experience = years
    def increase_salary(self,amount):
        self.salary = self.salary + amount
employee101 = Employee("Ali",101,"Karachi", 50000, "Manager",2020,"BSCS",5)
print(employee101.name)
print(employee101.number)
print(employee101.address)
print(employee101.salary)

employee101.describe_employee()
employee101.increase_salary(10000)
employee101.update_year_of_experience(6)

print("\nAfter Updating Employee101:")
employee101.describe_employee()

employee102 = Employee("Sara", 102, "Lahore", 45000, "Developer", 2021, "BSCS", 4)

employee103 = Employee("Ahmed", 103, "Karachi", 60000, "Designer", 2019, "BSIT", 6)

employee104 = Employee("Ayesha", 104, "Islamabad", 55000, "HR", 2022, "MBA", 3)

employee105 = Employee("Usman", 105, "Multan", 40000, "Accountant", 2023, "BBA", 2)

employee106 = Employee("Zara", 106, "Hyderabad", 70000, "Manager", 2018, "MBA", 8)

employee107 = Employee("Hamza", 107, "Karachi", 35000, "Assistant", 2024, "B.Com", 1)

employee108 = Employee("Hina", 108, "Lahore", 65000, "Developer", 2020, "BSCS", 5)

employee109 = Employee("Bilal", 109, "Quetta", 50000, "Analyst", 2021, "BSIT", 4)

employee110 = Employee("Noor", 110, "Sukkur", 48000, "Designer", 2022, "BSCS", 3)

employees = [employee101,employee102,employee103,employee104,employee105,employee106,
            employee107,employee108,employee109,employee110]
for emp in employees:
    emp.describe_employee()

employee102.increase_salary(5000)
employee103.update_year_of_experience(7)
employee104.increase_salary(10000)
employee105.update_year_of_experience(3)

print("\nAfter Some Updates:")

employee102.describe_employee()
employee103.describe_employee()
employee104.describe_employee()
employee105.describe_employee()
```
